### Core Mathematical Methodology
1. **Independent Per-Task Linear Probes**: Fits $\ell_2$-regularized logistic regression per task at Layer 25 ($l/L \approx 0.78$) to identify the unit truth direction $\hat{v} = \mathbf{w} / \|\mathbf{w}\|_2$.
2. **Horizontal Coordinate ($X$)**: Projections onto normalized truth direction: $x = (\mathbf{X} - \mu) \cdot \hat{v}$.
3. **Vertical Coordinate ($Y$)**: PCA fitted strictly to the orthogonal complement: $\mathbf{X}_{\perp} = (\mathbf{X} - \mu) - x \hat{v}^T$.
4. **Full Distribution Plotting**: Combined train and test distributions revealing geometric manifolds (F0 V-shape, F1 diagonal streaks, F2 crescent curve, F5 linear collapse vs CoT separation).
