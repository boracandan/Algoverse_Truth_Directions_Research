```python
import os

save_dir = "/content/drive/MyDrive/algoverse_results"
DATASETS_ROOT = "/content/Algoverse_Truth_Directions_Research/datasets"
os.makedirs(save_dir, exist_ok=True)

weights_path = f"{save_dir}/probe_weights.pkl"
cosine_csv_path = "/content/Algoverse_Truth_Directions_Research/experiments/cosine_similarity_mahalanobis.csv"

MODEL_NAME = "deepseek-ai/DeepSeek-R1-Distill-Llama-8B"
SHORT_MODEL_NAME = "deepseek-r1-distill-8b"
```

```python
# Loads the model ONCE, then sweeps every condition (no-prompt, no-prompt-chat-template,
# cot-zero-shot, sentence-based-CoT) x all 9 tasks x all layers, retraining each probe and
# saving its weight vector into weights_path. Replaces running all 4 separate notebooks.
compute_and_save_weights(
    model_name=MODEL_NAME,
    weights_path=weights_path,
    datasets_root=DATASETS_ROOT,
    short_model_name=SHORT_MODEL_NAME,
)
```

```python
# Merges the freshly computed weight vectors into the existing cosine-similarity CSV,
# adding weight_i/weight_j columns onto the matching (condition, layer, task_i, task_j) rows.
db = load_weights(weights_path)
append_weight_vectors(cosine_csv_path, db, model=SHORT_MODEL_NAME)
```
